      so when we try to do higher degree power multiplication using the inputs given in the birth rate graph 
      it becomes 
      (1902)^10 which is a huge number, same goes with other numbers
      and as polynomial regression uses least square solution , whose formula is 

      l=(x^TX)^-1X^Ty
      here The elements of design matrix X becomes very large and matrix inverse becomes very difficult as it makes X^TX ill conditioned.
      small numerical erros can make a huge error in the estimated coefficient.
      the possible solutions are minmax scalling in which we bring the inputs in the range of 0-1, whitening in which we make the mean 0, and unit variance and remove the corelation between the features.
